```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/Merged file(Solcast-NASA).csv')
```

```python
df.index = pd.to_datetime(df['timestamp'], format= '%d-%m-%Y %H:%M')
df.head()
```

```python
fig, ax = plt.subplots(figsize=(15, 4))
ax.plot(df.index, df['NASA_T'],  label='NASA')
ax.plot(df.index, df['Solcast_T'],  label='Solcast')
ax.set_xlim(pd.Timestamp('2016-01-01'), pd.Timestamp('2016-09-01'))
ax.set_ylabel('Ambient temperature $(^o$C)', fontsize=15, weight='bold')
ax.legend(loc='upper left')
ax.set_title('Ambient temperature of NASA vs Solcast data', weight='bold', fontsize=25)
ax.set_ylim(10, 45)
```
