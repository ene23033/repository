```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df=pd.read_csv('/content/Merged_NASA_Sol_WS.csv')
df.head()
```

```python
df.index = pd.to_datetime(df['timestamp'], format= '%d-%m-%Y %H:%M')
df.head()
```

```python
fig, ax = plt.subplots(figsize=(15, 4))
ax.plot(df.index, df['NASA_Temp'],  label='NASA')
ax.plot(df.index, df['Solcast_Temp'], label='Solcast')
ax.plot(df.index, df['Station_Temp'], label='Weather Station')
ax.set_xlim(pd.Timestamp('2024-12-24'), pd.Timestamp('2025-01-29'))
ax.set_ylabel('Ambient Temperature $(^o$C)', fontsize=12, weight='bold')
ax.legend(loc='upper left')
ax.set_title('24$^{th}$ Dec 2024 to 27$^{th}$ Jan 2025', weight='bold', fontsize=17)
ax.set_ylim(12, 40)
```

```python
fig, ax = plt.subplots(figsize=(15, 4))
ax.plot(df.index, df['NASA_Temp'],  label='NASA')
ax.plot(df.index, df['Solcast_Temp'], label='Solcast')
ax.plot(df.index, df['Station_Temp'], label='Weather Station')
ax.set_xlim(pd.Timestamp('2025-01-01'), pd.Timestamp('2025-01-02'))
ax.set_ylabel('Ambient Temperature $(^o$C)', fontsize=12, weight='bold')
ax.legend(loc='upper left')
ax.set_title('1$^{st}$ Jan 2025', weight='bold', fontsize=17)
ax.set_ylim(12, 40)
```

```python
fig,ax = plt.subplots()
ax.plot(df['Station_Temp'], df['NASA_Temp'], 'o', markersize='3')
ax.set_ylabel('NASA Ambient Temperature ($^o$C)', weight='bold', fontsize=12)
ax.set_xlabel('Weather Station Ambient Temperature ($^o$C)', weight='bold', fontsize=12)
plt.ylim(15,38)
plt.xlim(15,38)
```

```python
df['Station_Temp'].corr(df['NASA_Temp'])
```
