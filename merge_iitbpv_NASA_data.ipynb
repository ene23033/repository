```python
import pandas as pd
```

```python
df1 = pd.read_csv('/content/IITB-NASA data.csv')

df1.index = pd.to_datetime(df1['timestamp'], format='%d-%m-%Y %H:%M')
df1.head()
```

```python
df2 = pd.read_csv('/content/Final SCADA data(ground).csv')

df2.index = pd.to_datetime(df2['timestamp'], format='%m-%d-%Y %H:%M')
df2.head()
```

```python
df3 = pd.merge(df1, df2, left_on=df1.index, right_on=df2.index, how='inner')

df3 = df3.drop(columns=['key_0', 'timestamp_y'], axis=1)
df3 = df3.rename(columns={'timestamp_x':'timestamp'})

df3.to_csv('Merged file(ground POA-NASA).csv', index=False)

df3.tail()
```
