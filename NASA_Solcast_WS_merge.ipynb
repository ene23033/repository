```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df1=pd.read_csv('/content/NASA(24-01-24 to 27-01-25)one hr.csv')
df1.index=pd.to_datetime(df1['timestamp'], format= '%d-%m-%Y %H:%M')
df1.head()
```

```python
df2 = pd.read_csv('/content/weatherstationIITB(24-01-24 to 30-01-25) one min.csv')
df2.index = pd.to_datetime(df2['TIMESTAMP'], format='%d-%m-%Y %H:%M')
df2.head()
```

```python
df3 = pd.merge(df1, df2, left_on=df1.index, right_on=df2.index, how='inner')

df3.to_csv('AirT_Merged_file(NASA-WS).csv', index=False)

df3.head()
```
