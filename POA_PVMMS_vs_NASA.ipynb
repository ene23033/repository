```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/NASA+ground POA.csv')

df.index = pd.to_datetime(df['timestamp'], format='%d-%m-%Y %H:%M')
df['h'] = pd.to_datetime(df['timestamp'], format='%d-%m-%Y %H:%M').dt.hour
df.head()
```

```python
fig, ax = plt.subplots(figsize=(15, 4))
ax.plot(df.index, df['poa'], 'o', markersize=3, label='on-site POA')
ax.plot(df.index, df['NASA_poa'], 'o', markersize=3, label='NASA POA')
ax.set_xlim(pd.Timestamp('2016-01-15'), pd.Timestamp('2016-01-30'))
ax.set_ylabel('POA Irradiance (W/m$^2$)', fontsize=15, weight='bold')
ax.legend(loc='upper left')
ax.set_title('15$^{th}$ - 30$^{th}$ January 2016', weight='bold', fontsize=25)
ax.set_ylim(-50, 1200)
```

```python
fig, ax = plt.subplots(figsize=(15, 4))
ax.plot(df.index, df['poa'], 'o', markersize=3, label='on-site POA')
ax.plot(df.index, df['NASA_poa'], 'o', markersize=3, label='NASA POA')
ax.set_xlim(pd.Timestamp('2016-08-15'), pd.Timestamp('2016-08-30'))
ax.set_ylabel('POA Irradiance (W/m$^2$)', fontsize=15, weight='bold')
ax.set_title('15$^{th}$ - 30$^{th}$ August 2016', weight='bold', fontsize=25)
ax.legend(loc='upper left')
ax.set_ylim(-50, 1200)
```

```python
fig, ax = plt.subplots(figsize=(15, 4))
ax.plot(df.index, df['poa'], 'o', markersize=3, label='on-site POA')
ax.plot(df.index, df['NASA_poa'], 'o', markersize=3, label='NASA POA')
ax.set_xlim(pd.Timestamp('2017-03-1'), pd.Timestamp('2017-03-15'))
ax.set_ylabel('POA Irradiance (W/m$^2$)', fontsize=15, weight='bold')
ax.set_title('1$^{st}$ - 15$^{th}$ March 2017', weight='bold', fontsize=25)
ax.legend(loc='upper left')
ax.set_ylim(-50, 1200)
```

```python
s = pd.DataFrame(df[(df['h']>=8) & (df['h']<=17) & (df['poa']<1500)])

s.head()
```

```python
fig,ax = plt.subplots()
ax.plot(s['poa'], s['NASA_poa'], 'o', markersize='3')
ax.set_ylabel('NASA POA (W/m$^2$)', weight='bold', fontsize=15)
ax.set_xlabel('On-site POA (W/m$^2$)', weight='bold', fontsize=15)
```
