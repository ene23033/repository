```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pvlib
data= pd.read_csv("/content/Merged file(ground POA-NASA).csv")
```

```python
latitude = 19.1331  #  (Nagpur, latitude and longitude)
longitude = 72.9151
tilt = 19
azimuth = 180
from pvlib.location import Location
location = Location(latitude, longitude)
timestamp_index = pd.date_range(start="2015-03-01 00:00", periods=len(data), freq="h")
data.index = timestamp_index
```

```python
# Get solar position for the timestamps
solar_position = location.get_solarposition(timestamp_index)
```

```python
# Calculate POA irradiance
irradiance = pvlib.irradiance.get_total_irradiance(
    surface_tilt=tilt,
    surface_azimuth=azimuth,
    dni=data["DNI"],
    ghi=data["GHI"],
    dhi=data["DHI"],
    solar_zenith=solar_position["zenith"],
    solar_azimuth=solar_position["azimuth"],)
```

```python
# Extract POA irradiance
data["POA_irradiance"] = irradiance["poa_global"]
# Define a panel efficiency (20% as an example)
efficiency = 0.2

# Calculate solar generation as normalized output (p.u.)
data["Solar_Generation_pu"] = data["POA_irradiance"] * efficiency

# #Normalize generation between 0 and 1 (optional, already in p.u.)
# data["Solar_Generation_pu"] /= solar_data["Solar_Generation_pu"].max()
```

```python
output_path = "/content/solar_poa_IITB_PVMMS_NASA.xlsx"
data[["Solar_Generation_pu"]].to_excel(output_path)

output_path
```
